### Slide 9: Problem 2, a duplicate row
**Find it:**

```python
df.duplicated().sum()
```

**Fix it:**

```python
df = df.drop_duplicates()
```

### Slide 10: Problem 3, one label with different spellings
**Find it:** 5 "departments"?

```python
df["Dept"].unique()
```

**Fix it:** make every label capital letters.

```python
df["Dept"] = df["Dept"].str.upper()
```

### Slide 11: Problem 4, an impossible value
**Find it:**

```python
df["CGPA"].max()
```

**Fix it, step 1:** ask every row "is your CGPA at most 4.0?"

```python
df["CGPA"] <= 4.0
```

**Fix it, step 2:** keep only the True rows. 7 rows become 6.

```python
df = df[df["CGPA"] <= 4.0]
df.shape
```

## The cleaning recipe

| The job | Find it | Fix it |
|---|---|---|
| Open and look | `pd.read_csv("file.csv")`, `df.head()`, `df.shape`, `df.info()` | |
| Missing values | `df.isnull().sum()` | `df.dropna()` or `df["col"].fillna(value)` |
| Duplicate rows | `df.duplicated().sum()` | `df.drop_duplicates()` |
| Inconsistent labels | `df["col"].unique()` | `df["col"].str.upper()` |
| Impossible values | `df["col"].max()` | `df[df["col"] <= limit]` |

**Clean first. Then trust the answer.**
